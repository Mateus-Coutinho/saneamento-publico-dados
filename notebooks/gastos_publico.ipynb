```python
import pandas as pd

df = pd.read_csv('../data/ripsa011rc.csv').sort_values('Ano')

df.info()
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(df["Ano"], df["Numerador - Gasto público c/saneamento básico"])

plt.title('Gasto público com saneamento básico')
plt.xlabel('Ano')
plt.ylabel('Gasto público c/saneamento básico')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
